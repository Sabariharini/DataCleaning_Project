```python
for column in ['math score', 'reading score', 'writing score']:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print("Subject:", column)
    print("Lower limit:", lower)
    print("Upper limit:", upper)
    print("Number of outliers:", len(outliers))
    print("-------------------------")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('StudentsPerformance.csv')

df.head()
```

```python
for column in ['math score', 'reading score', 'writing score']:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print("\nSubject:", column)
    print(outliers[column].values)
```

```python
for column in ['math score', 'reading score', 'writing score']:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    print("\nOutlier records in:", column)

    display(df[df[column] < lower])
```

```python
# Create a copy of the original dataset
cleaned_df = df.copy()

# Check the cleaned dataset
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", cleaned_df.shape)
```

```python
cleaned_df.info()
```

```python
score_columns = ['math score', 'reading score', 'writing score']

for column in score_columns:
    print(column)
    print("Minimum:", cleaned_df[column].min())
    print("Maximum:", cleaned_df[column].max())
    print("-------------------")
```

```python
categorical_columns = [
    'gender',
    'race/ethnicity',
    'parental level of education',
    'lunch',
    'test preparation course'
]

for column in categorical_columns:
    print("\n", column)
    print(cleaned_df[column].unique())
```

```python
cleaned_df.to_csv('Cleaned_StudentsPerformance.csv', index=False)

print("Cleaned dataset saved successfully!")
```

```python
# Calculate correlation between student scores

correlation = cleaned_df[
    ['math score', 'reading score', 'writing score']
].corr()

print(correlation)
```

```python
# Create correlation heatmap

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title("Correlation Between Student Scores")
plt.show()
```
